```python
import pandas as pd

# Load Roller Coaster dataset
df = pd.read_csv("coaster_db.csv")

# Display first 5 rows
df.head()
```

```python
# Dataset shape
print("Dataset Shape:", df.shape)

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Data types and non-null values
print("\nDataset Information:")
df.info()

# Missing values
print("\nMissing Values:")
print(df.isnull().sum().sort_values(ascending=False).head(20))

# Statistical summary
print("\nStatistical Summary:")
df.describe(include="all").T
```

```python
# Check dataset dimensions
print("Rows and Columns:", df.shape)

# Missing values percentage
missing_data = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_data = missing_data.sort_values(
    by="Missing Values",
    ascending=False
)

missing_data.head(20)
```

```python
# Display all column names with their data types

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column} → {df[column].dtype}")
```

```python
# Create a copy of the original dataset
eda_df = df.copy()

# Remove duplicate rows
print("Duplicates before cleaning:", eda_df.duplicated().sum())

eda_df = eda_df.drop_duplicates()

print("Duplicates after cleaning:", eda_df.duplicated().sum())

# Select useful columns for analysis
selected_columns = [
    "coaster_name",
    "Location",
    "Status",
    "Type_Main",
    "Manufacturer",
    "year_introduced",
    "speed_mph",
    "height_ft",
    "Inversions_clean",
    "Gforce_clean",
    "latitude",
    "longitude"
]

eda_df = eda_df[selected_columns].copy()

# Convert important columns to numeric
numeric_columns = [
    "year_introduced",
    "speed_mph",
    "height_ft",
    "Inversions_clean",
    "Gforce_clean",
    "latitude",
    "longitude"
]

for column in numeric_columns:
    eda_df[column] = pd.to_numeric(
        eda_df[column],
        errors="coerce"
    )

# Display the cleaned dataset
print("Cleaned dataset shape:", eda_df.shape)

eda_df.head()
```

```python
# Step 3: Missing Value Analysis

print("Missing Values:")
print(eda_df.isnull().sum().sort_values(ascending=False))

# Missing value percentage
missing_percentage = (
    eda_df.isnull().sum() / len(eda_df) * 100
).sort_values(ascending=False)

print("\nMissing Value Percentage:")
print(missing_percentage)

# Statistical summary of numerical columns
print("\nStatistical Summary:")
eda_df.describe().T
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count roller coasters by status
status_counts = eda_df["Status"].value_counts()

print(status_counts)

# Create visualization
plt.figure(figsize=(10, 6))

sns.barplot(
    x=status_counts.index,
    y=status_counts.values
)

plt.title("Roller Coaster Status Distribution")
plt.xlabel("Status")
plt.ylabel("Number of Roller Coasters")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Count the top 10 manufacturers
manufacturer_counts = (
    eda_df["Manufacturer"]
    .dropna()
    .value_counts()
    .head(10)
)

print(manufacturer_counts)

# Create visualization
plt.figure(figsize=(12, 6))

sns.barplot(
    x=manufacturer_counts.values,
    y=manufacturer_counts.index
)

plt.title("Top 10 Roller Coaster Manufacturers")
plt.xlabel("Number of Roller Coasters")
plt.ylabel("Manufacturer")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    eda_df["speed_mph"].dropna(),
    bins=30,
    kde=True
)

plt.title("Distribution of Roller Coaster Speeds")
plt.xlabel("Speed (mph)")
plt.ylabel("Number of Roller Coasters")

plt.tight_layout()
plt.show()
```

```python
speed_height = eda_df[
    ["speed_mph", "height_ft"]
].dropna()

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=speed_height,
    x="height_ft",
    y="speed_mph",
    alpha=0.7
)

plt.title("Relationship Between Height and Speed")
plt.xlabel("Height (ft)")
plt.ylabel("Speed (mph)")

plt.tight_layout()
plt.show()
```

```python
corr_columns = [
    "year_introduced",
    "speed_mph",
    "height_ft",
    "Inversions_clean",
    "Gforce_clean",
    "latitude",
    "longitude"
]

corr_matrix = eda_df[corr_columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    center=0
)

plt.title("Correlation Heatmap of Numerical Features")
plt.tight_layout()
plt.show()
```

```python
location_counts = (
    eda_df["Location"]
    .dropna()
    .value_counts()
    .head(10)
)

print(location_counts)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=location_counts.values,
    y=location_counts.index
)

plt.title("Top 10 Locations by Number of Roller Coasters")
plt.xlabel("Number of Roller Coasters")
plt.ylabel("Location")

plt.tight_layout()
plt.show()
```

```python
top_types = (
    eda_df["Type_Main"]
    .value_counts()
    .head(8)
    .index
)

plot_df = eda_df[
    eda_df["Type_Main"].isin(top_types)
][["Type_Main", "speed_mph"]].dropna()

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=plot_df,
    x="Type_Main",
    y="speed_mph"
)

plt.title("Speed Distribution by Roller Coaster Type")
plt.xlabel("Main Type")
plt.ylabel("Speed (mph)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
type_summary = (
    eda_df.groupby("Type_Main")["speed_mph"]
    .agg(["count", "mean", "median", "min", "max"])
    .sort_values("mean", ascending=False)
)

type_summary.head(10)
```

```python
year_summary = (
    eda_df.dropna(subset=["year_introduced", "speed_mph"])
    .groupby("year_introduced")["speed_mph"]
    .mean()
    .reset_index()
    .sort_values("year_introduced")
)

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=year_summary,
    x="year_introduced",
    y="speed_mph",
    marker="o"
)

plt.title("Average Roller Coaster Speed by Introduction Year")
plt.xlabel("Year Introduced")
plt.ylabel("Average Speed (mph)")

plt.tight_layout()
plt.show()
```

```python
top_fastest = (
    eda_df[
        [
            "coaster_name",
            "Location",
            "Manufacturer",
            "speed_mph",
            "height_ft"
        ]
    ]
    .dropna(subset=["speed_mph"])
    .sort_values("speed_mph", ascending=False)
    .head(10)
)

top_fastest
```

```python
# Generate important findings from the dataset

print("===== KEY FINDINGS =====")

print("\n1. Dataset Overview")
print(f"Total roller coasters: {len(eda_df)}")

print("\n2. Most Common Status")
print(eda_df["Status"].value_counts().head(3))

print("\n3. Top Manufacturers")
print(eda_df["Manufacturer"].value_counts().head(5))

print("\n4. Speed Statistics")
print(eda_df["speed_mph"].describe())

print("\n5. Height Statistics")
print(eda_df["height_ft"].describe())

print("\n6. Fastest Roller Coaster")
fastest = eda_df.loc[
    eda_df["speed_mph"].idxmax()
]
print(fastest[[
    "coaster_name",
    "Manufacturer",
    "speed_mph",
    "height_ft"
]])

print("\n7. Correlation: Height vs Speed")
print(
    eda_df["height_ft"].corr(
        eda_df["speed_mph"]
    )
)
```
